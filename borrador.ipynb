### Notas sobre **StockCode**

El campo `StockCode` **debería** identificar un producto físico con un código numérico de 5 dígitos  
(ej.: `85123`). Sin embargo, la base trae varios valores “especiales” que NO representan
productos sino cargos administrativos, descuentos o ajustes.  
Para evitar que ensucien las métricas RFM (sobre todo _Monetary_), conviene filtrarlos
antes del clustering.

| Código               | Qué significa, en criollo                                                                        | ¿Lo usamos en el clustering? | Motivo                                |
| -------------------- | ------------------------------------------------------------------------------------------------ | ---------------------------- | ------------------------------------- |
| `POST` / `DOT`       | Gastos de correo / flete que la empresa le pasa al cliente.                                      | **Excluir**                  | No es compra real.                    |
| `D`                  | Descuento global aplicado a la factura.                                                          | **Excluir**                  | Valor negativo, no es ítem.           |
| `C2`                 | “Carriage” (otro tipo de cargo de transporte).                                                   | **Excluir**                  | Sin unidades físicas.                 |
| `M` o `m`            | Movimiento manual cargado por un admin (ajuste contable).                                        | **Excluir**                  | No refleja comportamiento de compra.  |
| `BANK CHARGES` / `B` | Comisiones bancarias repercutidas al cliente.                                                    | **Excluir**                  | Gasto financiero.                     |
| `S`                  | Muestras gratuitas enviadas al cliente.                                                          | **Excluir**                  | No hay pago asociado.                 |
| `AMAZONFEE`          | Comisión cobrada por ventas despachadas a través de Amazon.                                      | **Excluir**                  | Gasto logístico.                      |
| `DCGSXXXX`           | “Discount Gift Set”: kits promocionales; muchos vienen con cantidad negativa y sin `CustomerID`. | **Excluir**                  | Datos incompletos y valores atípicos. |
| `gift_0001_10/20…`   | Ítems pagados con gift-card; no traen datos de cliente y alteran el gasto.                       | **Excluir**                  | No aportan al análisis de lealtad.    |
| `CRUK`               | Donación a la organización “Cancer Research UK” incluida en el ticket.                           | **Excluir**                  | Donación, no producto.                |
| `PADS`               | Material de embalaje real (padding).                                                             | **Incluir**                  | Producto físico con venta legítima.   |

> **Observación:** Nos quedamos sólo con los `StockCode` que representen **productos físicos vendidos**.
> Los otros códigos distorsionan el gasto total y la frecuencia sin aportar
> información sobre el comportamiento de compra del cliente.

Conservamos el código PADS porque corresponde a un artículo físico —material de embalaje o «padding»— que el cliente efectivamente paga: aparece con cantidades positivas, precio unitario y siempre asociado a un CustomerID. No es un cargo administrativo ni un ajuste contable; forma parte de la cesta de compra real. Excluirlo restaría ventas legítimas al cálculo de Monetary y disminuiría la Frequency, subestimando el verdadero valor y hábito de compra del cliente.

Calcular la actualidad, la frecuencia y los valores monetarios:

Actualidad (R) como días desde la última compra: ¿Cuántos días hace de su última compra? Reste la fecha de compra más reciente de hoy para calcular el valor de actualidad. ¿Hace 1 día? ¿Hace 14 días? ¿Hace 500 días?

Monetario (M) como dinero total gastado: ¿Cuánto $$ (o la moneda que utilice para calcularlo) ha gastado este cliente? Nuevamente, limite el gasto a los últimos dos años o tome todo el tiempo. Simplemente sume el dinero de todas las transacciones para obtener el valor M.

# Análisis RFM


```python
aggregated_df = cleaned_df.groupby(by="CustomerID", as_index=False).agg(
    MonetaryValue=("revenue", "sum"),
    Frequency=("InvoiceNo", "nunique"),
    LastInvoiceDate=("InvoiceDate", "max"),
)

aggregated_df.head(5)
```

```python
max_invoice_date = aggregated_df["LastInvoiceDate"].max()

aggregated_df["Recency"] = (max_invoice_date - aggregated_df["LastInvoiceDate"]).dt.days

aggregated_df.head(5)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.hist(aggregated_df["MonetaryValue"], bins=10, color="skyblue", edgecolor="black")
plt.title("Monetary Value Distribution")
plt.xlabel("Monetary Value")
plt.ylabel("Count")

plt.subplot(1, 3, 2)
plt.hist(aggregated_df["Frequency"], bins=10, color="lightgreen", edgecolor="black")
plt.title("Frequency Distribution")
plt.xlabel("Frequency")
plt.ylabel("Count")

plt.subplot(1, 3, 3)
plt.hist(aggregated_df["Recency"], bins=20, color="salmon", edgecolor="black")
plt.title("Recency Distribution")
plt.xlabel("Recency")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

La primera serie de histogramas describe con nitidez un patrón muy concentrado y fuertemente sesgado a la derecha en las tres variables RFM.

Monetary Value: la mayor parte de los clientes gasta montos modestos—se apilan cerca de cero—mientras que existe una cola larga que llega a valores superiores a $250 000.

Frequency: algo semejante; el 75 % compra entre 1 y 5 veces, pero aparecen casos excepcionales que superan las 200 facturas.

Recency: el pico inicial indica muchos clientes que compraron hace poco (recency baja), y la cola se extiende hasta ~360 días, señal de cuentas prácticamente inactivas.
Este comportamiento revela una clientela típicamente “long-tail”: pocos “ballenas” y muchísimos compradores pequeños.


```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.boxplot(data=aggregated_df["MonetaryValue"], color="skyblue")
plt.title("Monetary Value Boxplot")
plt.xlabel("Monetary Value")

plt.subplot(1, 3, 2)
sns.boxplot(data=aggregated_df["Frequency"], color="lightgreen")
plt.title("Frequency Boxplot")
plt.xlabel("Frequency")

plt.subplot(1, 3, 3)
sns.boxplot(data=aggregated_df["Recency"], color="salmon")
plt.title("Recency Boxplot")
plt.xlabel("Recency")

plt.tight_layout()
plt.show()
```

Las tres variables RFM presentan la típica “cola larga” del comercio electrónico.

Monetary Value y Frequency: la inmensa mayoría de los clientes gasta poco y compra entre 1 y 5 veces; solo un puñado llega a montos y frecuencias muy altos.

Recency: muchos compradores recientes (valores cercanos a 0 días) y una cola que se estira hasta el año, señal de cuentas casi inactivas.

Los boxplots lo dejan en evidencia: el rectángulo —que abarca el 50 % central de los datos— es muy bajo, mientras que los “bigotes” son cortos y seguidos por una nube de puntos aislados (outliers) que se disparan en las tres métricas. Esto confirma que esos pocos clientes extremos podrían sesgar promedios y desviar algoritmos de clustering basados en distancia.


```python
M_Q1 = aggregated_df["MonetaryValue"].quantile(0.25)
M_Q3 = aggregated_df["MonetaryValue"].quantile(0.75)
M_IQR = M_Q3 - M_Q1

monetary_outliers_df = aggregated_df[
    (aggregated_df["MonetaryValue"] > (M_Q3 + 1.5 * M_IQR))
    | (aggregated_df["MonetaryValue"] < (M_Q1 - 1.5 * M_IQR))
].copy()

monetary_outliers_df.describe()
```

En esta parte calculamos los puntos que dividen en cuatro cuartiles la distribución de MonetaryValue y a partir de ellos definimos el rango intercuartílico (IQR), una medida robusta de dispersión que ignora los valores más extremos. Con Q1 y Q3 en la mano, establecemos un umbral de 1,5 veces el IQR por encima y por debajo de ese rango central, según la regla clásica de Tukey.

A continuación, filtramos el DataFrame original para quedarnos sólo con los clientes que superan esos límites y así aislar los casos realmente atípicos en sus montos de compra. El resultado se copia en monetary_outliers_df, que contiene exclusivamente esos outliers.


```python
F_Q1 = aggregated_df["Frequency"].quantile(0.25)
F_Q3 = aggregated_df["Frequency"].quantile(0.75)
F_IQR = F_Q3 - F_Q1

frequency_outliers_df = aggregated_df[
    (aggregated_df["Frequency"] > (F_Q3 + 1.5 * F_IQR))
    | (aggregated_df["Frequency"] < (F_Q1 - 1.5 * F_IQR))
].copy()

frequency_outliers_df.describe()
```

En este bloque determinamos los cuartiles primero (F_Q1 y F_Q3) de la métrica Frequency y calculamos su rango intercuartílico (IQR) como la diferencia entre ambos. Ese IQR funciona como medida de dispersión central y nos permite definir un umbral razonable para la mayoría de los datos sin que influyan los casos extremos.

Con esos umbrales en mano, filtramos el DataFrame aggregated_df para quedarnos solo con los clientes cuya frecuencia de compra caiga por debajo de F_Q1 − 1.5·IQR o supere F_Q3 + 1.5·IQR. De ese modo aislamos los outliers de frecuencia y los guardamos en frequency_outliers_df.


```python
non_outliers_df = aggregated_df[
    (~aggregated_df.index.isin(monetary_outliers_df.index))
    & (~aggregated_df.index.isin(frequency_outliers_df.index))
]

non_outliers_df.describe()
```

En esta parte se construye un nuevo DataFrame que agrupa sólo a los clientes “normales”, excluyendo a quienes ya identificamos como outliers en Monetary y en Frequency. Para ello, utilizamos index.isin() sobre los índices de los DataFrames de outliers y aplicamos la negación (~) para quedarnos con los registros que no están en ninguna de esas dos listas.

Finalmente, al ejecutar non_outliers_df.describe(), obtenemos un resumen estadístico de este subconjunto libre de atípicos. De esa manera podemos comparar la media, la dispersión y los rangos de la mayoría de los clientes sin que los valores extremos distorsionen las métricas.


```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.boxplot(data=non_outliers_df["MonetaryValue"], color="skyblue")
plt.title("Monetary Value Boxplot")
plt.xlabel("Monetary Value")

plt.subplot(1, 3, 2)
sns.boxplot(data=non_outliers_df["Frequency"], color="lightgreen")
plt.title("Frequency Boxplot")
plt.xlabel("Frequency")

plt.subplot(1, 3, 3)
sns.boxplot(data=non_outliers_df["Recency"], color="salmon")
plt.title("Recency Boxplot")
plt.xlabel("Recency")

plt.tight_layout()
plt.show()
```

En Frequency, el IQR va de 1 a 4 compras por cliente, con mediana en 2, y apenas sobreviven algunos valores atípicos hasta 11 facturas. La dispersión reducida confirma que el comportamiento habitual se concentra en muy pocas transacciones por cliente.

$$

Donde:

- $z$ es el valor estandarizado,
- $x$ es el valor original,
- $\mu$ es la media de la feature,
- $\sigma$ es la desviación estandar de la feature.

# K-MEANS CLUSTERING


K-Means intenta separar los datos en grupos con formas circulares similares, minimizando la distancia interna en cada grupo. Usamos varias métricas para evaluar qué número de grupos da la mejor segmentación, y lo visualizamos en un espacio reducido (PCA) para ver si los grupos realmente se separan de forma clara.


```python
# ---------- K-MEANS ----------
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    calinski_harabasz_score,
    davies_bouldin_score,
)
from sklearn.decomposition import PCA

k_range = range(2, 11)
sil_scores_kmeans, ch_scores_kmeans, db_scores_kmeans, inertia = [], [], [], []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(scaled_data_df)
    sil_scores_kmeans.append(silhouette_score(scaled_data_df, labels))
    ch_scores_kmeans.append(calinski_harabasz_score(scaled_data_df, labels))
    db_scores_kmeans.append(davies_bouldin_score(scaled_data_df, labels))
    inertia.append(kmeans.inertia_)

# Visualización de métricas
plt.figure(figsize=(14, 4))
plt.subplot(1, 4, 1)
plt.plot(k_range, inertia, marker="o")
plt.title("Inertia")
plt.subplot(1, 4, 2)
plt.plot(k_range, sil_scores_kmeans, marker="o")
plt.title("Silhouette")
plt.subplot(1, 4, 3)
plt.plot(k_range, ch_scores_kmeans, marker="o")
plt.title("Calinski-Harabasz")
plt.subplot(1, 4, 4)
plt.plot(k_range, db_scores_kmeans, marker="o")
plt.title("Davies-Bouldin")
plt.suptitle("KMeans - Métricas")
plt.tight_layout()
plt.show()

# Clustering final con el mejor k
best_k_kmeans = k_range[sil_scores_kmeans.index(max(sil_scores_kmeans))]
kmeans_final = KMeans(n_clusters=best_k_kmeans, random_state=42)
labels_kmeans = kmeans_final.fit_predict(scaled_data_df)

# Visualización PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data_df)
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=pca_result[:, 0], y=pca_result[:, 1], hue=labels_kmeans, palette="tab10", s=60
)
plt.title(f"KMeans (k={best_k_kmeans}) - PCA")
plt.show()

# Imprimir métricas finales
print(f"KMEANS - Silhouette: {max(sil_scores_kmeans):.4f}")
print(f"KMEANS - CH: {max(ch_scores_kmeans):.2f}")
print(f"KMEANS - DB: {min(db_scores_kmeans):.4f}")
print(f"KMEANS - Clusters Detectados: {best_k_kmeans}")
```

El método K-Means mostró un desempeño muy sólido en este análisis. Se detectaron 3 clusters como partición óptima, confirmado por un valor Silhouette de 0.4609, que indica que los grupos están relativamente bien definidos y separados entre sí. La métrica Calinski-Harabasz alcanzó 4279.60, lo cual es un valor muy alto y demuestra una fuerte cohesión interna dentro de cada grupo y buena separación entre ellos. Además, el índice Davies-Bouldin fue de 0.7601, lo cual también es favorable (valores cercanos a cero indican menos solapamiento entre clusters). Esta combinación de métricas y la visualización mediante PCA refuerzan la idea de que K-Means logró una segmentación efectiva y bien estructurada de los clientes en base a sus comportamientos RFM.


# AGGLOMERATIVE CLUSTERING


Agglomerative funciona como un árbol que agrupa datos de abajo hacia arriba. El dendrograma nos ayuda a ver cómo se construyen los grupos. Luego probamos distintas cantidades de clusters para ver cuál tiene mejor cohesión y separación.


```python
# ---------- AGGLOMERATIVE ----------
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# Dendrograma
plt.figure(figsize=(12, 5))
Z = linkage(scaled_data_df, method="ward")
dendrogram(Z, truncate_mode="lastp", p=30, leaf_rotation=45.0, leaf_font_size=10.0)
plt.title("Dendrograma - Agglomerative")
plt.xlabel("Clientes")
plt.ylabel("Distancia")
plt.show()

# Métricas
k_range = range(2, 11)
sil_scores_agg, ch_scores_agg, db_scores_agg = [], [], []

for k in k_range:
    agg = AgglomerativeClustering(n_clusters=k, linkage="ward")
    labels = agg.fit_predict(scaled_data_df)
    sil_scores_agg.append(silhouette_score(scaled_data_df, labels))
    ch_scores_agg.append(calinski_harabasz_score(scaled_data_df, labels))
    db_scores_agg.append(davies_bouldin_score(scaled_data_df, labels))

# Gráficos de métricas
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1)
plt.plot(k_range, sil_scores_agg, marker="o")
plt.title("Silhouette")
plt.subplot(1, 3, 2)
plt.plot(k_range, ch_scores_agg, marker="o")
plt.title("Calinski-Harabasz")
plt.subplot(1, 3, 3)
plt.plot(k_range, db_scores_agg, marker="o")
plt.title("Davies-Bouldin")
plt.suptitle("Agglomerative - Métricas")
plt.tight_layout()
plt.show()

# Mejor k
best_k_agg = k_range[sil_scores_agg.index(max(sil_scores_agg))]
agg_final = AgglomerativeClustering(n_clusters=best_k_agg, linkage="ward")
labels_agg = agg_final.fit_predict(scaled_data_df)

# Visualización PCA
pca_result = PCA(n_components=2).fit_transform(scaled_data_df)
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=pca_result[:, 0], y=pca_result[:, 1], hue=labels_agg, palette="tab10", s=60
)
plt.title(f"Agglomerative (k={best_k_agg}) - PCA")
plt.show()

# Imprimir métricas
print(f"AGGLOMERATIVE - Silhouette: {max(sil_scores_agg):.4f}")
print(f"AGGLOMERATIVE - CH: {max(ch_scores_agg):.2f}")
print(f"AGGLOMERATIVE - DB: {min(db_scores_agg):.4f}")
print(f"AGGLOMERATIVE - Clusters Detectados: {best_k_agg}")
```

El modelo Agglomerative identificó 2 clusters como la mejor partición según las métricas evaluadas. Su Silhouette alcanzó 0.4808, siendo el valor más alto de todos los métodos aplicados, lo cual sugiere que la agrupación lograda es especialmente compacta y con buena separación entre los grupos. La métrica Calinski-Harabasz fue de 3106.16, también muy elevada, lo que respalda la estabilidad estructural de los clusters. Por otro lado, Davies-Bouldin fue de 0.6979, el valor más bajo de todos los modelos, lo que indica menor solapamiento entre grupos. Si bien el número de clusters detectados fue menor que en K-Means, la calidad interna de los mismos fue incluso superior. Esto posiciona a Agglomerative como un modelo de alta calidad, ideal para segmentaciones donde la jerarquía y estructura natural entre clientes puede jugar un rol clave.


# GAUSSIAN MIXTURE MODEL


GMM asume que los datos provienen de varias “nubes” con forma de elipse (distribuciones normales). Esto permite más flexibilidad que K-Means. Se evalúan las métricas con distintos grupos y se proyectan en 2D para entender qué tan bien se separan.


```python
# ---------- GAUSSIAN MIXTURE ----------
from sklearn.mixture import GaussianMixture

k_range = range(2, 11)
sil_scores_gmm, ch_scores_gmm, db_scores_gmm = [], [], []

for k in k_range:
    gmm = GaussianMixture(n_components=k, random_state=42)
    gmm.fit(scaled_data_df)
    labels = gmm.predict(scaled_data_df)
    sil_scores_gmm.append(silhouette_score(scaled_data_df, labels))
    ch_scores_gmm.append(calinski_harabasz_score(scaled_data_df, labels))
    db_scores_gmm.append(davies_bouldin_score(scaled_data_df, labels))

# Métricas
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1)
plt.plot(k_range, sil_scores_gmm, marker="o")
plt.title("Silhouette")
plt.subplot(1, 3, 2)
plt.plot(k_range, ch_scores_gmm, marker="o")
plt.title("Calinski-Harabasz")
plt.subplot(1, 3, 3)
plt.plot(k_range, db_scores_gmm, marker="o")
plt.title("Davies-Bouldin")
plt.suptitle("GMM - Métricas")
plt.tight_layout()
plt.show()

# Mejor k
best_k_gmm = k_range[sil_scores_gmm.index(max(sil_scores_gmm))]
gmm_final = GaussianMixture(n_components=best_k_gmm, random_state=42)
labels_gmm = gmm_final.fit_predict(scaled_data_df)

# Visualización PCA
pca_result = PCA(n_components=2).fit_transform(scaled_data_df)
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=pca_result[:, 0], y=pca_result[:, 1], hue=labels_gmm, palette="tab10", s=60
)
plt.title(f"GMM (k={best_k_gmm}) - PCA")
plt.show()

# Imprimir métricas
print(f"GMM - Silhouette: {max(sil_scores_gmm):.4f}")
print(f"GMM - CH: {max(ch_scores_gmm):.2f}")
print(f"GMM - DB: {min(db_scores_gmm):.4f}")
print(f"GMM - Clusters Detectados: {best_k_gmm}")
```

El modelo Gaussian Mixture detectó 2 clusters, con métricas algo más débiles que los modelos anteriores. La Silhouette fue de 0.3334, lo que indica que los puntos dentro de un mismo cluster no están tan bien separados entre sí, y que puede haber mayor solapamiento entre grupos. La métrica Calinski-Harabasz fue de 2388.67, considerable pero inferior respecto a K-Means y Agglomerative. El índice Davies-Bouldin alcanzó 1.1400, más alto que el resto, lo que sugiere mayor confusión entre los clusters. Aunque GMM es útil cuando se espera solapamiento o se busca una segmentación más suave (probabilística), en este caso su rendimiento fue menor. Aun así, puede ser valioso como complemento o para escenarios donde las diferencias entre grupos no son tan claras.


# DBSCAN


DBSCAN encuentra grupos sin necesidad de definir cuántos clusters queremos. Detecta zonas densas de clientes y marca como “ruido” los que están muy separados. Es útil para identificar anomalías o clientes aislados.


```python
# ---------- DBSCAN ----------
from sklearn.cluster import DBSCAN
from sklearn.metrics import (
    silhouette_score,
    calinski_harabasz_score,
    davies_bouldin_score,
)

eps = 0.9
min_samples = 5
db = DBSCAN(eps=eps, min_samples=min_samples).fit(scaled_data_df)
labels_dbscan = db.labels_

# Validar si hay al menos 2 clusters
if len(set(labels_dbscan)) > 1:
    sil_dbscan = silhouette_score(scaled_data_df, labels_dbscan)
    ch_dbscan = calinski_harabasz_score(scaled_data_df, labels_dbscan)
    dbi_dbscan = davies_bouldin_score(scaled_data_df, labels_dbscan)
    n_clusters_dbscan = len(set(labels_dbscan)) - (1 if -1 in labels_dbscan else 0)

    # Visualización PCA
    pca_result = PCA(n_components=2).fit_transform(scaled_data_df)
    plt.figure(figsize=(8, 6))
    sns.scatterplot(
        x=pca_result[:, 0], y=pca_result[:, 1], hue=labels_dbscan, palette="tab10", s=60
    )
    plt.title(f"DBSCAN (eps={eps}, min_samples={min_samples}) - PCA")
    plt.show()

    print(f"DBSCAN - Silhouette: {sil_dbscan:.4f}")
    print(f"DBSCAN - CH: {ch_dbscan:.2f}")
    print(f"DBSCAN - DB: {dbi_dbscan:.4f}")
    print(f"DBSCAN - Clusters Detectados: {n_clusters_dbscan}")
else:
    print("DBSCAN no detectó suficientes clusters. Ajustar eps o min_samples.")
```

DBSCAN no logró una segmentación efectiva en este caso, ya que detectó solamente 1 cluster. Si bien identificó una zona densa de clientes, no logró diferenciar grupos internos dentro del conjunto de datos. Las métricas reflejan esta limitación: la Silhouette fue de 0.4059, moderada, pero no muy significativa dado que no hubo separación real de clusters. El Calinski-Harabasz fue apenas 18.05, lo que muestra una cohesión interna débil y escasa separación. El índice Davies-Bouldin fue de 1.1592, el más alto de todos los modelos, indicando solapamiento significativo. En resumen, aunque DBSCAN puede ser útil para detectar outliers o anomalías, en este caso no generó una segmentación válida y no es recomendable como modelo principal para agrupar a los clientes según sus características RFM.

# Interpretación y Recomendaciones


A partir de los resultados obtenidos con el algoritmo K-Means y el análisis visual mediante los violin plots de Recency, Frequency y Monetary Value por cluster, es posible etiquetar y perfilar de forma clara los tres grupos de clientes formados. Se trata de los compradores más activos, fieles y rentables. Son clientes que la empresa debe retener y fidelizar con acciones especiales, ya que representan el núcleo fuerte del negocio.

El Cluster 1 incluye a clientes que también tienen recencia baja (lo cual indica que han comprado recientemente), pero presentan niveles bajos tanto de frecuencia como de valor monetario. Alternativamente, pueden dejar de formar parte del foco comercial si los recursos son limitados y se prioriza rentabilidad sobre volumen.


## 🔵 Cluster 0 – Clientes estratégicos


El objetivo principal con este grupo es mantener su fidelidad y seguir fortaleciendo el vínculo comercial, ya que representan el segmento más rentable y estable del negocio. Si bien su comportamiento ya es positivo, siempre existe el riesgo de que dejen de comprar si no se sienten valorados o si encuentran una mejor propuesta en la competencia. Es fundamental sostener su interés sin caer en descuentos innecesarios que erosionen el margen.

Acciones de marketing recomendadas:

Propuestas personalizadas en base a historial de compra. Analizar qué productos suelen adquirir estos clientes y ofrecerles recomendaciones bien curadas de artículos complementarios o nuevas líneas relacionadas. Esto no sólo eleva el ticket promedio, sino que además posiciona a la empresa como conocedora de sus intereses, fortaleciendo la percepción de cercanía y profesionalismo.

Comunicación diferenciada desde un canal directo. En lugar de correos masivos, se sugiere utilizar una casilla específica del equipo de ventas para contactarlos de forma personalizada cada cierto tiempo (por ejemplo, una vez por trimestre), agradeciendo su continuidad, consultando necesidades puntuales o anticipando lanzamientos de temporada.

Acceso prioritario a condiciones especiales. Darles la posibilidad de comprar productos nuevos antes de su lanzamiento general o de reservar stock limitado con prioridad. Estas acciones refuerzan el valor simbólico de pertenecer a una “categoría premium” de clientes y reducen el riesgo de desinterés por saturación de campañas genéricas.

Programas de fidelización progresiva. No necesariamente con un esquema formal de puntos, pero sí con beneficios acumulables (envíos sin cargo a partir de cierto monto, bonificaciones a fin de trimestre, etc.), que se comuniquen de forma clara y que muestren cómo su lealtad tiene una recompensa concreta.

Revisión y seguimiento activo de experiencia de compra. Se recomienda contactar a estos clientes al menos una vez al año para pedir su opinión sobre la atención, los productos o los envíos, y utilizar esa información para ajustar procesos. Esto no sólo mejora el servicio, sino que demuestra que su voz es tenida en cuenta.


## 🟡 Cluster 1 – Clientes de potencial medio


Este grupo tiene un gran valor latente: son clientes que ingresaron recientemente, realizaron alguna compra, pero aún no repiten ni generan un vínculo sostenido. Es probable que hayan tenido una experiencia correcta, pero no lo suficientemente destacada como para incorporarlos al hábito de compra. El foco con ellos debe ser educar, acompañar y facilitar una segunda compra que marque la diferencia.

Acciones de marketing recomendadas:

Diseñar una secuencia de bienvenida guiada. Al primer contacto, enviar una campaña segmentada de tres correos en una semana con información útil: sugerencias de productos, tips de uso, beneficios de comprar en la tienda y una promoción exclusiva para su segunda compra. Esto reduce la fricción y genera familiaridad con la marca.

Ofrecer beneficios reales por recompra temprana. Por ejemplo, un descuento fijo ($1000 o un producto bonificado) si realizan un segundo pedido dentro de los 15 días posteriores al primero. Es fundamental que el beneficio sea concreto, de aplicación fácil y sin letras chicas, para generar confianza y urgencia razonable.

Mostrarles productos populares dentro de su categoría. No sobrecargarlos con todo el catálogo. Si compraron artículos de papelería, por ejemplo, enviar un listado de los más vendidos en esa línea. La claridad del mensaje es clave para que exploren sin esfuerzo.

Incluir casos reales o testimonios. Compartir ejemplos de cómo otros clientes armaron sus pedidos, organizaron regalos o aprovecharon descuentos. Esto genera pertenencia y reduce la inseguridad en la decisión de compra, sobre todo si son compradores nuevos o de bajo involucramiento.

Reforzar el canal de atención. Asegurarse de que estos clientes sepan cómo contactarse ante dudas o problemas. Muchas veces, no repiten por temor a equivocarse o porque tuvieron una consulta sin respuesta. Un mail breve post-compra ofreciendo ayuda puede marcar la diferencia.


## 🔴 Cluster 2 – Clientes inactivos


Estos clientes han dejado de comprar desde hace un tiempo considerable. No necesariamente están disconformes, pero su comportamiento indica desinterés o cambio de prioridades. El objetivo principal con este grupo es revisar si pueden reactivarse con acciones puntuales, y si no, entender las causas del abandono para ajustar las estrategias a futuro.

Acciones de marketing recomendadas:

Envío de packs prearmados con propuesta clara. No se trata de ofrecer todo el catálogo, sino de presentar uno o dos combos cerrados con alto valor percibido, a buen precio y listos para comprar en un clic. Esto evita que el cliente tenga que navegar o tomar muchas decisiones, lo cual muchas veces es una barrera en clientes inactivos.

Encuestas de abandono cortas y bien dirigidas. Enviar una consulta simple con tres opciones (“dejé de comprar por precio”, “porque no encontré variedad”, “porque me resultó difícil el proceso”) y un campo abierto opcional. Este feedback es clave para detectar patrones y ajustar el enfoque comercial. Ofrecer un pequeño beneficio a quienes respondan puede mejorar la tasa de retorno.

Contacto personalizado para clientes inactivos de alto valor. Si dentro de este cluster hay clientes que alguna vez fueron rentables, es necesario ir más allá del email. Se recomienda contactarlos por teléfono o con una carta directa, preguntar si algo cambió y ofrecer una propuesta adaptada a sus necesidades. Esta atención diferencial puede ser decisiva para recuperar cuentas importantes.

Revisión de datos de contacto. Muchas veces el abandono no se debe a la falta de interés, sino a errores en los medios de contacto (emails mal escritos, cambios de casilla, etc.). Se sugiere revisar esta información o solicitar un segundo medio de contacto en el proceso de registro para evitar pérdidas futuras.

Definir un umbral de abandono para no insistir. Si luego de dos o tres intentos espaciados el cliente no responde, conviene dejar de dirigirle campañas activas y asignarlo a una lista pasiva que reciba solo comunicaciones generales en fechas clave, sin invertir más recursos en su seguimiento.
