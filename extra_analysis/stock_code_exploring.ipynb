# Exploring StockCode


Vamos a realizar en esta notebook la exploración de los "StockCode" porque sino nos va agregar ruido innecesario a nuestro EDA.


Al final vamos a incluir un resumen sobre qué vamos a hacer con cada variedad, para definir si las incluimos en el análisis o si es mejor descartarlas.


```python
df["StockCode"] = df["StockCode"].astype(str)
df["InvoiceNo"] = df["InvoiceNo"].astype(str)
```

```python
df[~df["StockCode"].str.match(r"^\d{5}[a-zA-Z]*$")]["StockCode"].unique()
```

```python
df[df["StockCode"].str.startswith("POST")]
```

A primera vista, parece el costo asociado al envío. Queremos ver si se trata de una factura común y corriente, es decir, que tiene otras compras en la misma.

Por eso veamos un caso en particular, por ejemplo el primero.


```python
df[df["InvoiceNo"] == "536370"]
```

Es el costo asociado al envío.


```python
df[df["StockCode"] == "D"]
```

Parecen ser lineas de ventas destinadas a representar un descuento, por eso la cantidad es negativa. Vemos si realmente esto se da siempre.


```python
df[df["StockCode"] == "D"]["Quantity"].unique()
```

Tal parece que sí. Vamos a excluir de nuestro análisis los descuentos.


```python
df[df["StockCode"] == "C2"].head(10)
```

```python
df[df["StockCode"] == "C2"]["Description"].unique()
```

Este tipo de registros también parecen asociados a envíos. Vamos a descartarlos.


```python
df[df["StockCode"] == "DOT"].head(10)
```

Estos también parecen estar relacionados con costos de envío. Y vemos que de la muestra de 10 que tomamos, ninguno tiene un "CustomerID" asociado. Al igual que con los demás, vamos a descartar estos.


```python
df[df["StockCode"] == "M"].head(10)
```

```python
df[df["StockCode"] == "M"].describe()
```

```python
df[df["InvoiceNo"] == "536981"].head(10)
```

Este caso es más complicado. Parece que se tratan de cargas en facturas que se hacen manualmente.

Esto puede ser debido a que no tienen un código asociado a ese producto en ese momento, o cuando se trata de ajustes de stock.

Son pocos, alrededor de 500 registros, en primera instancia creemos que deberíamos incluirlos porque afectan al total de la factura. Pero ante la duda, y siendo que son pocos, preferimos descartarlos.


```python
df[df["StockCode"] == "BANK CHARGES"].head(10)
```

```python
df[df["StockCode"] == "BANK CHARGES"].describe()
```

Parece ser una linea para indicar costos adicionales relacionados al banco. Son 37 casos únicamente. Y tan solo 12 de ellos tienen un "CustomerID" asociado. Vamos a descartarlos.


```python
df[df["StockCode"] == "S"].head(10)
```

Parecen ser muestras gratis de productos dados a los clientes.


Podrían ser importantes sin en efecto lo que hacen es anular el precio de productos en las facturas de clientes. Pero son pocos casos. Veamos un poco más:


```python
df[df["StockCode"] == "S"].describe()
```

Ninguno de estos registros de muestra gratis tiene un "CustomerID" asociado. Vamos a descartarlos.


```python
df[df["StockCode"] == "AMAZONFEE"].head(10)
```

```python
df[df["StockCode"] == "AMAZONFEE"].describe()
```

Parecen ser tarifas cobradas por Amazon por ventas por medio de su plataforma. Sea lo que sea, son tan solo 34 registros y ninguno tiene un "CustomerID" asignado. Vamos a descartar esto.


```python
df[df["StockCode"].str.startswith("DCGS")].describe()
```

De todos estos "StockCode" que comienzan con DCGS. Ninguno tiene un "CustomerID" asociado. No nos sirven. Vamos a descartarlos.


```python
df[df["StockCode"].str.startswith("gift")].describe()
```

De vuelta lo mismo. Los que empiezan con "gift" parecerían indicar regalos. Pero de todas formas ninguno tiene un "CustomerID" así que los descartamos.


```python
df[df["StockCode"] == "m"]
```

Era de estas cargas manuales, solo que con la m minúscula. También lo descartamos. Es uno solo y no tiene un "CustomerID" asociado.


```python
df[df["StockCode"] == "PADS"]
```

Esto, traducido, parece tratarse de almohadillas o rellenos que combinan con todos los cojines. Son 4 registros con este "StockCode". Y parecen válidos. Con un "UnitPrice" muy bajo. Pero sin embargo no dejan de ser válidos. Vamos a dejarlos.


```python
df[df["StockCode"] == "B"]
```

Parecen ser ajustes contables. Los vamos a descartar.


```python
df[df["StockCode"] == "CRUK"]
```

Según internet. CRUK puede referirse a una organización benéfica de UK dedicada a la investigación del cáncer. Una 'commission' podría hacer referencia a donaciones o comisiones destinadas a esta organización.


No representa un consumo que nos sea pertinente para el análisis de clientes, por lo que lo vamos a excluir.


Con esto damos por terminado el análisis de StockCode.
